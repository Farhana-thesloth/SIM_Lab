```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
raw_random_data = np.random.rand(30000)
normal_day_waits = np.random.uniform(2, 12, 3000)
peak_day_waits = np.random.normal(8, 2, 3000)
peak_day_waits = np.clip(peak_day_waits, 0, None)

print("Regular days : ", normal_day_waits)
print("Busy day : ", peak_day_waits)
fig, plot_axes = plt.subplots(1, 2, figsize=(15, 5))

plot_axes[0].hist(
    normal_day_waits, bins=40, alpha=0.4, color="blue", edgecolor="black"
)
plot_axes[0].set_title("Regular Day Wait Times Uniform")
plot_axes[0].set_xlabel("Wait Time minutes")
plot_axes[0].set_ylabel("Frequency")
plot_axes[0].grid(True)

plot_axes[1].hist(
    peak_day_waits, bins=50, alpha=0.5, color="red", edgecolor="black"
)
plot_axes[1].set_title("Busy Day Wait Times Normal")
plot_axes[1].set_xlabel("Wait Time minutes")
plot_axes[1].set_ylabel("Frequency")
plot_axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
avg_normal = np.mean(normal_day_waits)
med_normal = np.median(normal_day_waits)
std_normal = np.std(normal_day_waits)
max_normal = np.max(normal_day_waits)

# Busy day calculations
avg_peak = np.mean(peak_day_waits)
med_peak = np.median(peak_day_waits)
std_peak = np.std(peak_day_waits)
max_peak = np.max(peak_day_waits)

print("\nRegular Day :")
print("Mean:", avg_normal)
print("Median:", med_normal)
print("Standard Deviation:", std_normal)
print("Maximum Wait Time:", max_normal)

print("\nBusy Day :")
print("Mean:", avg_peak)
print("Median:", med_peak)
print("Standard Deviation:", std_peak)
print("Maximum Wait Time:", max_peak)
```
