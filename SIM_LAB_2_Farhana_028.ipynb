# 1.How did the determinant change and why?

 Think of the determinant for a $2 \times 2$ matrix as $ad - bc$ (top-left $\times$ bottom-right minus top-right $\times$ bottom-left).For $A$: $(10 \times 4) - (3 \times 16) = 40 - 48 = -8$For $B$: $(10 \times 4) - (4 \times 16) = 40 - 64 = -24$By increasing the top-right number from 3 to 4, you increased the subtraction part ($bc$), pulling the determinant further into the negative. Geometric-wise, the determinant represents the "area scaling factor" (and orientation flip, due to the negative sign) of the space transformed by the matrix. Stretching that component enlarged the area transformation factor by 3 times.

# 2. Did the rank change? Explain.

No, the rank stayed at 2.Why: Rank simply measures how many independent directions (or non-redundant rows/columns) a matrix has. For a $2 \times 2$ matrix, the rank only drops to 1 if one row is an exact multiple of the other, which makes the determinant 0. Since both determinants are non-zero ($-8$ and $-24$), both matrices keep full dimensional capability (rank 2). Neither row is a copy or simple scale of the other.
# 3. How did the eigenvalues respond to the value change?

The eigenvalues shifted from roughly [14.55, -0.55] in $A$ to [15.54, -1.54] in $B$.Why: Eigenvalues are the scale factors along special directional axes. Two quick checks tie them back to matrix properties:Sum (Trace): The sum of the diagonal numbers ($10 + 4 = 14$) must equal the sum of the eigenvalues. Notice that $14.55 + (-0.55) = 14$, and $15.54 + (-1.54) = 14$. The diagonal sum didn't change, so the sum of the eigenvalues couldn't change either!Product (Determinant): Multiplying the eigenvalues gives the determinant. Since the determinant dropped from $-8$ to $-24$, the two eigenvalues had to spread further apart—one growing more positive and the other growing more negative—to multiply out to $-24$ while keeping their sum at $14$.
# 4. Is B easier or harder to invert than A? Why?

Mathematically, they are equally easy.Why: Both are simple $2 \times 2$ matrices, so computing the inverse uses the exact same step-by-step formula.Numerically: $B$ is actually slightly better/more stable to invert than $A$. When a matrix's determinant gets very close to 0, it becomes "almost singular," making small computational errors blow up during inversion. Since $B$ has a determinant ($-24$) further away from 0 than $A$ ($-8$), $B$ is farther from being singular, making its inverse safer and less sensitive to rounding errors in computer arithmetic.
