```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
scalar = 3.7
arr = np.array([5,9,3,4,1,9,4])

sinV = np.sin(arr)
cosV = np.cos(arr)
tanV = np.tan(arr)
asinV = np.arcsin(np.clip(arr / np.max(arr), -1, 1))
atanV = np.arctan(arr)


expV = np.exp(arr)
logV = np.log(arr)


absV = np.abs(arr)
sqrtV = np.sqrt(arr)


remV = np.remainder(arr, 2)


roundV = np.round(arr)
floorV = np.floor(arr)
ceilV = np.ceil(arr)

print("sin:", sinV)
print("cos:", cosV)
print("tan:", tanV)
print("asin:", asinV)
print("atan:", atanV)
print("exp:", expV)
print("log:", logV)
print("abs:", absV)
print("sqrt:", sqrtV)
print("rem:", remV)
print("round:", roundV)
print("floor:", floorV)
print("ceil:", ceilV)
```

```python
M = np.random.uniform(1.0, 10.0, size=(2, 3))
maxV = np.max(M)
max_index = np.argmax(M)

minV = np.min(M)
min_index = np.argmin(M)

M_length = len(M)

sorted_rows_desc = np.flip(np.sort(M, axis=1), axis=1)
sorted_entire_desc = np.flip(np.sort(M.flatten()))
#Flattening means taking a multi-dimensional array (like a 2D matrix with rows and columns) and squishing it down into a single 1D line of numbers.

sumV = np.sum(M)
prodV = np.prod(M)
medianV = np.median(M)
meanV = np.mean(M)
std_dev = np.std(M)
print("Matrix:\n", M)
print("Max value:", maxV, "at index", max_index)
print("Min value:", minV, "at index", min_index)
print("Length of matrix:", M_length)
print("Sorted row-wise descending:\n", sorted_rows_desc)
print("Sorted entire matrix descending:", sorted_entire_desc)
print("Sum of elements:", sumV)
print("Product of elements:", prodV)
print("Median value:", medianV)
print("Mean value:", meanV)
print("Standard deviation:", std_dev)

sns.heatmap(M, annot=True, cmap="Blues")
plt.title("2*3 Matrix Heatmap")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.show()
```
