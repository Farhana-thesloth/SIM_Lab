```python
import math
import random
import numpy as np

# --- Part 1: Setup & Personalization ---
my_id = 28
r = my_id / 10  # Range [0, 2.8]
name = "Farhana"

# Count letters in name
letters = 0
for ch in name:
    if ch.isalpha():
        letters += 1

base_pts = 200 * letters

print("Range Used: [0, ", r, "]")
print("Number of letters in name:", letters)
print("Starting points:", base_pts)
print("\n" + "=" * 75 + "\n")


# --- Part 2: Simulation Logic ---
def in_circle(x, y, radius):
    return x * x + y * y <= radius * radius


def simulate(total_pts, radius):
    half = total_pts // 2
    inside_cnt = 0

    # First half using random.uniform()
    for i in range(half):
        x = random.uniform(0, radius)
        y = random.uniform(0, radius)
        if in_circle(x, y, radius):
            inside_cnt += 1

    # Second half using numpy.random.rand()
    x_arr = np.random.rand(total_pts - half) * radius
    y_arr = np.random.rand(total_pts - half) * radius

    for i in range(len(x_arr)):
        if in_circle(x_arr[i], y_arr[i], radius):
            inside_cnt += 1

    return inside_cnt


# --- Part 3: Output & Analysis ---
print(
    f"{'Range Used':<12} | {'Total Points':<12} | {'Points Inside':<14} |"
    f" {'Estimated Pi':<14} | {'Error'}"
)
print("-" * 75)

for i in range(1, 6):
    total = base_pts * i
    inside = simulate(total, r)

    est_pi = 4 * (inside / total)
    err = abs(math.pi - est_pi)

    print(
        f"[0, {r}]      | {total:<12} | {inside:<14} | {est_pi:<14.6f} |"
        f" {err:.6f}"
    )
```
